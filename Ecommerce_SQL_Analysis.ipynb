```python
#Create SQLite Database
import pandas as pd
df = pd.read_csv('data.csv', encoding='ISO-8859-1')

print(df.head())
```

```python
#SQL Queries
conn = sqlite3.connect(':memory:')  # Create an in-memory SQLite database
df.to_sql('ecommerce_data', conn, index=False, if_exists='replace') # Write DataFrame to SQL table

query = """
SELECT *
FROM ecommerce_data
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
#WHERE Query
query = """
SELECT *
FROM ecommerce_data
WHERE Quantity > 10;
"""

pd.read_sql(query, conn)
```

```python
#GROUP BY Query
query = """
SELECT Country, SUM(Quantity) AS Total_Quantity
FROM ecommerce_data
GROUP BY Country;
"""

pd.read_sql(query, conn)
```

```python
#ORDER BY Query
query = """
SELECT Description, UnitPrice
FROM ecommerce_data
ORDER BY UnitPrice DESC;
"""

pd.read_sql(query, conn)
```

```python
#AVG Function
query = """
SELECT AVG(UnitPrice) AS Average_Price
FROM ecommerce_data;
"""

pd.read_sql(query, conn)
```

```python
#Top Products
query = """
SELECT Description, SUM(Quantity) AS Total_Sales
FROM ecommerce_data
GROUP BY Description
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
#Subquery
query = """
SELECT *
FROM ecommerce_data
WHERE UnitPrice > (
    SELECT AVG(UnitPrice)
    FROM ecommerce_data
);
"""

pd.read_sql(query, conn)
```

```python
#Create View
query = """
CREATE VIEW high_sales_products AS
SELECT Description, SUM(Quantity) AS Total_Quantity
FROM ecommerce_data
GROUP BY Description;
"""

conn.execute(query)

print("VIEW CREATED SUCCESSFULLY")
```

```python
#NULL Handling
query = """
SELECT *
FROM ecommerce_data
WHERE CustomerID IS NULL;
"""

pd.read_sql(query, conn)
```

```python
#Create Index
query = """
CREATE INDEX idx_country
ON ecommerce_data(Country);
"""

conn.execute(query)

print("INDEX CREATED SUCCESSFULLY")
```
